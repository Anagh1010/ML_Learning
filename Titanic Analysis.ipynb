# Handling Missing Values

```python
df.head()
```

## 1. Drop missing values using dropna()

```python
df.shape
```

```python
df.dropna().shape
```

```python
df.dropna(axis=1).shape
```

```python
df.dropna(subset=['Cabin', 'Embarked']).shape
```

```python
# how is an important parameter. By default its 'any' meaning dropna will drop all those rows which have null value for atleast one column
# if you give how='all' then only those rows for which all columns are null
# Use inplace to make it permanent
```

```python
# another parameter is the axis. By default its 0 meaning rows. If axis = 1 then it will drop all columns which have null value for atleast one or all rows(depends on how parameter)
```

```python
# subset parameter can be used to give it a single column or a list of columns and dropna will check only those columns for null values
# if list is passed python will do an OR operation meaning null in either this or this or both then drop row(union operation)
```

## 2. Fill missing values using fillna()

```python
#Detecting Outliers code

q1=df['Age'].quantile(0.25)
q3=df['Age'].quantile(0.75)
q2=df['Age'].quantile(0.50) #Median
iqr=q3-q1 #inter-quantile range
min_threshold=q1-iqr*1.5
max_threshold=q3+iqr*1.5
outliers=df[(df['Age']<min_threshold) | (df['Age']>max_threshold)]
new_df=df[(df['Age']>min_threshold) & (df['Age']<max_threshold)]
```
